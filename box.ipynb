```python
def box(a):
    try:
        from tabulate import tabulate
        from IPython.display import HTML
        display(HTML(tabulate(a,'keys','html').replace('>\n','>').replace('\n','<br>')))
    except ModuleNotFoundError:
        a = {str(i):[str(a)for a in a]if not isinstance(a,str)and hasattr(a,'__iter__')else str(a).splitlines()for i,a in a.items()}
        l = {i:max(len(i),max(map(len,a)))for i,a in a.items()}
        t = []
        for i in a:
            t.append(i + (l[i]-len(i))*' ')
        print('│'.join(t))
        for n in range(max(map(len,a.values()))):
            t = []
            for i,j in a.items():
                j = j[n] if n<len(j) else ''
                t.append(j + (l[i]-len(j))*' ')
            print('│'.join(t))
```
